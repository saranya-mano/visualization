```python
#import libraries

import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

#set seed
np.random.seed(613)
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.set_style('darkgrid')

tipgraph = sns.lineplot(data = tips,
             x = 'total_bill',
             y = 'tip')

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')

plt.show()
```

```python
#create and customize plot

fig = plt.subplots(figsize=(10,3))

tipgraph = sns.lineplot(data = tips,
                        x = 'total_bill',
                        y = 'tip',
                        color = 'hotpink',
                        linestyle = '--',
                        linewidth = 3,
                        marker = 'o',
                        markerfacecolor = 'orange')

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
#customize multiple axes at once

tipgraph = sns.scatterplot(data = tips,
                           x = 'total_bill',
                           y = 'tip',
                           style = 'time',
                           hue = 'day', 
                           palette = ['purple', 'hotpink', 'deepskyblue', 'yellowgreen'])

tipgraph.set(title = 'Tips vs Total Bill',
             xlabel = 'Total Bill ($)',
             ylabel = 'Tip Amount ($)')
```

```python
sns.pairplot(
    data = tips,
    hue = 'day'
)
```

```python
# create multiple plots to explore data
daysplot = sns.relplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex',
    col = 'day',
    kind = 'scatter',
    col_wrap = 2
)
```

```python
#import libraries

import plotly.graph_objects as go 
import nbformat

#create dataset

x1 = np.array(['Luffy', 'Xoro', 'Zeno', 'Hugh'])
y1 = np.array([112, 333, 444, 534])
```

```python
#plot graph using graph objects

graph = go.Figure()

graph.add_trace(go.Bar(x = x1, y = y1))

graph.update_layout(
    title = "Pirate Scores",
    xaxis_title = "Pirates",
    yaxis_title = "Score"
)

graph.show()
```

```python
#customize plotly graphs

graph = go.Figure()
graph.add_trace(go.Scatter(
    x = x1, 
    y = y1,
    mode = 'markers',
    marker = dict(
        size = 15,
        color = 'hotpink',
        opacity = 1, #opacity is set to 100%
        line = dict(width = 2, color = 'purple')
    )
))

graph.update_layout(
    title = 'Interactive Pirate Plot',
    xaxis_title = 'Pirates',
    yaxis_title = 'Scores',
    width = 500, 
    height = 500
)
```

```python
#import Wordcloud module and read CSV file into dataframe

from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
                 on_bad_lines='skip')
df
```

```python
# join all text from the column quote into one string
text = " ".join(each for each in df.quote)

#generate an object called wordcloud 
wordcloud = WordCloud(background_color="white",
                      colormap= "inferno",
                      random_state= 613).generate(text)

# use matplotlib to add image to a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, 
          interpolation = 'bilinear') #helps smooth out image
ax.axis('off')
```

```python
# import packages
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
```

```python
# define sets
A = set(['apple', 'banana', 'watermelon'])
```

```python
B = set(['pumpkin', 'blueberry', 'apple', 'key lime'])

diagram = venn2([A, B],
                set_labels = ('Fruits', 'Pies'),
                set_colors = ('blue', 'red'),
                alpha = 0.5)

diagram.get_label_by_id('10').set_text('\n'.join(A-B)) ##binary numbers - left on, right off - Everything in A minus B
diagram.get_label_by_id('11').set_text('\n'.join(A&B)) ##binary numbers - middle slice - Everything in both A and B
diagram.get_label_by_id('01').set_text('\n'.join(B-A)) ##binary numbers - right - Everything in B minus A 
#list out all of these items separated by a new line
```
