```python
#import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#set seed
np.random.seed(613)

#create dataset
x = np.arange(50)
y = np.random.randint(0, 100, 50)

#create figure and axes and plot graph
fig, ax = plt.subplots(figsize=(5,3))
#ax.scatter(x, y)
#ax.bar(x, y)
ax.barh(x, y)
#ax.plot(x, y)
#ax.hist(y)

#add title and labels
ax.set_title('Total growth over time')
ax.set_xlabel('Something')
ax.set_ylabel('Somethign else')
```

```python
#create font families using dictionary
font1 = {'family': 'sans-serif', 'color': 'blue', 'size': 23}
font2 = {'family': 'monospace', 'color': 'red', 'size': 12}

#create graph
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)

#add title and labels and apply font style
ax.set_title('Total growth over time', fontdict = font1, loc = 'right')
ax.set_xlabel('Something', fontdict = font2)
ax.set_ylabel('Somethign else', fontdict = font2)

```

```python
#plot graph and customize markers
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y, 
           marker = '*', 
           markersize = 12,
           color = 'indigo',
           linestyle = '--',
             linewidth = 6,
              markeredgecolor = '#fa9359',
               markerfacecolor = '#000000') #difference between linewidth and linewidths

#add gridlines and customize them
#ax.grid() #for both axes
ax.grid(axis = 'x') #quotation marks are important
ax.grid(axis = 'y', linewidth = 5, color = 'hotpink', linestyle = '-.') #limited linestyles allowed, refer to guide
```
