```python
#import libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
#create data set
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)

#make a line plot with labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')

#add legend and customize it
ax.legend(loc = 'lower left',
          frameon = True, #add frame around the legend
          fontsize = 12, #change font size
          ncol = 1, #specific number of columns
          shadow = True,
          bbox_to_anchor = (1,0.75))
```

```python

#make a scatter plot with labels and legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc = 'upper right')

ax.text(10,95, "This value is not important",
        ha = 'left', #alignment of text
        color = 'red', #modify font colour
        size = 10) #modify font size
```

```python
#create plot with annotations 
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1, 5, ". Data:(1,5)", transform=ax.transData)
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure)
```

```python
#add arrows to annotations
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y1, label = "Person 1")
ax.scatter(x, y2, label = "Person 2")
ax.legend(loc = "lower right")

ax.annotate('This is not important', xy = (10, 95), xytext=(20, 94), arrowprops=dict(facecolor='hotpink')) 
```

```python
#create sample data
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Salt", "Oil"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# add subplots
fig, (ax1, ax2) = plt.subplots(ncols = 2,
                               nrows = 1,
                               figsize = (7,3))
ax1.scatter(x1, y1, label = "Dots")
ax2.bar(x2, y2, label = "Bars")
ax1.legend(loc = 'upper right')
ax2.legend(loc = 'upper right')

ax1.text(10, 95, "I am a bunch of dots")
ax2.text(.5, 200, "I am a ton of bars")
```

```python
#customize subplots
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    layout = "constrained") #helps ensure that legends/labels are constrained within the plot
someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

someaxes["ax1"].set_xlabel('A biiiig label', fontsize = 18)
someaxes["ax2"].set_xlabel('A smaaaaal label', fontsize = 18)
someaxes["ax3"].set_xlabel('Blah blah blah', fontsize = 18)
```

```python
#create sample data
x = np.array(["Luffy", "Zoro", "Nami", "Salt", "Oil"])
y1 = np.array([110, 170, 99, 120, 67])
y2 = np.array([170, 100, 90, 120, 50])

#add and customize subplots
fig, ax = plt.subplots(figsize = (7, 3))
ax.bar(x, y1, color = "indigo")
ax.plot(x, y2, color = "hotpink")
```

```python
#plot standard deviation and error bars
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x, y2, color = "red")

ax.errorbar(x, y2, yerr = y2_sd, fmt = "none")
```

```python
#import libraries

from PIL import Image
from io import BytesIO
```

```python
#add image from link

response = requests.get("https://static.wikia.nocookie.net/the-ultimate-good/images/3/3d/Luffy.png/revision/latest?cb=20230417200546")
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
#add image to plot

fig, ax = plt.subplots(figsize = (7, 3))
ax.plot(x, y2, color = "hotpink")

ax_image = fig.add_axes([0.1, 
                         0.11,
                         0.15,
                         0.35])

ax_image.imshow(image)
ax_image.axis('off')
```
